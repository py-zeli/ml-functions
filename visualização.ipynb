```python
import functions

x = [2, 3, 5, 7, 8]
y = [5, 7, 8, 10, 12]

evolucao = {
    "Iteração":[],
    "Coeficiente Linear": [],
    "Coeficiente Angular": [],

}

list_coef_linear = []

taxa_de_aprendizado = 0.02

for i in range(1500):

    if 'novo_coef_linear' in locals():
        valores_calibrados = functions.gradient_descent(x, y, taxa_de_aprendizado, novo_coef_linear, novo_coef_angular)
        
        evolucao["Coeficiente Linear"].append(valores_calibrados['novo_coef_linear'])
        evolucao["Coeficiente Angular"].append(valores_calibrados['novo_coef_angular'])
        
    
    else:
        valores_calibrados = functions.gradient_descent(x, y, taxa_de_aprendizado)
        evolucao["Coeficiente Linear"].append(valores_calibrados['novo_coef_linear'])
        evolucao["Coeficiente Angular"].append(valores_calibrados['novo_coef_angular'])
    
    evolucao["Iteração"].append(i+1)
    # print(f"Resultados da {i+1} iteração:\n{valores_calibrados}\n")
    # # valores_calibrados = print(functions.gradient_descent(x, y, 0.01))
    novo_coef_linear = valores_calibrados['novo_coef_linear']
    novo_coef_angular = valores_calibrados['novo_coef_angular']
```

```python
from matplotlib import pyplot as plt

plt.figure(figsize=(10,5))
plt.title("Evolução dos Coeficientes")
plt.xlabel("Iteração")
plt.ylabel("Valor encontrado")
plt.plot(evolucao["Iteração"], evolucao["Coeficiente Linear"], color = "green", label = "Coeficiente Linear")
plt.plot(evolucao["Iteração"], evolucao["Coeficiente Angular"], color = "red", label = "Coeficiente Angular")
plt.legend(loc='upper left')
plt.show()
```
